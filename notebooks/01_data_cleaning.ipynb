## 6. Converter e validar datas

Converta `data_pedido` para um tipo de data consistente.

Existe pelo menos um valor que não representa uma data válida. Não force uma correção sem evidência.

**Funções/métodos úteis:**
- `pd.to_datetime()`
- parâmetro `errors="coerce"`
- parâmetro `dayfirst=True`
- `.isna()`


```python
datas_raw = df_clean["data_pedido"].copy()
datas_numericas = pd.to_numeric(datas_raw, errors="coerce")
mask_excel = datas_numericas.notna()
df_clean.loc[mask_excel, ["pedido_id", "data_pedido"]].head(10)
```

```python
datas_convertidas = pd.Series(
    pd.NaT,
    index=df_clean.index,
    dtype="datetime64[ns]"
)
datas_convertidas.loc[mask_excel] = pd.to_datetime(
    datas_numericas.loc[mask_excel],
    unit="D",
    origin="1899-12-30"
)
pd.DataFrame({
    "original": datas_raw[mask_excel].head(),
    "convertida": datas_convertidas[mask_excel].head()
})
```

```python
mask_texto = ~mask_excel

datas_texto = (
    datas_raw.loc[mask_texto]
    .astype("string")
    .str.strip()
)

# Formato DD/MM/YYYY
datas_br_barra = pd.to_datetime(
    datas_texto,
    format="%d/%m/%Y",
    errors="coerce"
)

# Formato YYYY-MM-DD
datas_iso = pd.to_datetime(
    datas_texto,
    format="%Y-%m-%d",
    errors="coerce"
)

# Formato DD-MM-YYYY
datas_br_hifen = pd.to_datetime(
    datas_texto,
    format="%d-%m-%Y",
    errors="coerce"
)

# Formato DD/MM/YY
datas_br_curto = pd.to_datetime(
    datas_texto,
    format="%d/%m/%y",
    errors="coerce"
)
```

```python
datas_texto_convertidas = (
    datas_br_barra
    .fillna(datas_iso)
    .fillna(datas_br_hifen)
    .fillna(datas_br_curto)
)

datas_convertidas.loc[mask_texto] = datas_texto_convertidas
df_clean["data_pedido"] = datas_convertidas
```

## 7. Converter campos numéricos

Trate:
- `quantidade`
- `preco_unitario`
- `desconto_pct`

A planilha mistura números reais e textos formatados.

Antes da conversão, talvez seja necessário remover símbolos, `%`, `R$`, separadores ou espaços.

**Funções/métodos úteis:**
- `.str.replace()`
- `pd.to_numeric()`
- `errors="coerce"`
- `.astype()`

**Atenção:** vírgula pode aparecer como separador decimal em alguns registros.


```python
df_clean["preco_unitario"].unique()
```

```python
preco_limpo = (
    df_clean["preco_unitario"]
    .astype("string")
    .str.strip()
    .str.replace("R$", "", regex=False)
    .str.replace(",", ".", regex=False)
)

preco_convertido = pd.to_numeric(preco_limpo, errors="coerce")

df_clean.loc[
    preco_convertido.isna(),
    ["pedido_id", "preco_unitario",]
]
```

```python
df_clean["preco_unitario"] = preco_convertido
```

```python
df_clean["desconto_pct"].unique()
```

```python
df_clean["desconto_pct"] = df_clean["desconto_pct"].replace({
    "0%": 0,
    "15%": 15,
    "5%": 5,
    "10%": 10,
    "cinco": 5
})
df_clean["desconto_pct"].unique()
```

```python
# Emails ausentes
emails_ausentes = df_clean["email"].isna().sum()

# Emails inválidos, excluindo os ausentes
padrao_email = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

emails_invalidos = (
    df_clean["email"].notna()
    & ~df_clean["email"].str.match(padrao_email)
).sum()

# Estados inválidos
ufs_validas = [
    "AC", "AL", "AP", "AM", "BA", "CE", "DF", "ES", "GO",
    "MA", "MT", "MS", "MG", "PA", "PB", "PR", "PE", "PI",
    "RJ", "RN", "RS", "RO", "RR", "SC", "SP", "SE", "TO"
]

estados_invalidos = (~df_clean["estado"].isin(ufs_validas)).sum()

# Quantidades inválidas ou ausentes
quantidades_invalidas = (
    df_clean["quantidade"].isna()
    | (df_clean["quantidade"] <= 0)
).sum()

# Descontos ausentes
descontos_ausentes = df_clean["desconto_pct"].isna().sum()

# Preços ausentes ou inválidos
precos_invalidos = (
    df_clean["preco_unitario"].isna()
    | (df_clean["preco_unitario"] <= 0)
).sum()
```

## 9. Tratar valores ausentes e anomalias

Separe três situações:
1. valor que pode ser corrigido de forma determinística;
2. valor que deve permanecer ausente;
3. registro que merece ser sinalizado para revisão.

Não use `fillna()` indiscriminadamente.

**Funções/métodos úteis:**
- `.isna()`
- `.loc[]`
- `.fillna()` quando houver justificativa
- criação de uma coluna de controle, como `revisar`


```python
padrao_email = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

emails_problematicos = (
    df_clean["email"].isna()
    | ~df_clean["email"].str.match(padrao_email, na=False)
)

estados_problematicos = ~df_clean["estado"].isin(ufs_validas)

quantidades_problematicas = (
    df_clean["quantidade"].isna()
    | (df_clean["quantidade"] <= 0)
)

precos_problematicos = (
    df_clean["preco_unitario"].isna()
    | (df_clean["preco_unitario"] <= 0)
)

descontos_problematicos = df_clean["desconto_pct"].isna()

datas_problematicas = df_clean["data_pedido"].isna()
```

```python
df_clean["revisar"] = (
    emails_problematicos
    | estados_problematicos
    | quantidades_problematicas
    | precos_problematicos
    | descontos_problematicos
    | datas_problematicas
    | ids_ausentes
    | ids_duplicados
)
```

```python
df_clean["revisar"].value_counts()
```

```python
df_clean["revisar"].sum()
```

```python
df_clean[df_clean["revisar"]]
```

```python
df_clean["analise"] = ""

df_clean.loc[emails_problematicos, "analise"] += "email; "
df_clean.loc[estados_problematicos, "analise"] += "estado; "
df_clean.loc[quantidades_problematicas, "analise"] += "quantidade; "
df_clean.loc[precos_problematicos, "analise"] += "preco_unitario; "
df_clean.loc[descontos_problematicos, "analise"] += "desconto; "
df_clean.loc[datas_problematicas, "analise"] += "data; "
df_clean.loc[ids_ausentes, "analise"] += "pedido_id ausente; "
df_clean.loc[ids_duplicados, "analise"] += "pedido_id duplicado; "

df_clean["analise"] = df_clean["analise"].str.rstrip("; ")
```

## 10. Criar campos derivados

Crie pelo menos uma coluna útil a partir dos dados limpos.

Sugestão:
- calcular `valor_bruto = quantidade × preco_unitario`;
- calcular `valor_liquido` aplicando `desconto_pct`.

**Funções/métodos úteis:**
- operações vetorizadas entre colunas;
- `.round()`


```python
df_clean["valor_bruto"] = (
    df_clean["quantidade"] * df_clean["preco_unitario"]
)

df_clean["valor_liquido"] = (
    df_clean["valor_bruto"] * (1 - df_clean["desconto_pct"] / 100)
)

df_clean["valor_bruto"] = df_clean["valor_bruto"].round(2)
df_clean["valor_liquido"] = df_clean["valor_liquido"].round(2)
```

A limpeza removeu duplicatas exatas, padronizou campos textuais, converteu datas e valores numéricos, reduziu a coluna categoria de 15 representações para 4 categorias canônicas e sinalizou 33 registros para revisão manual. Valores ambíguos não foram corrigidos automaticamente.

```python
def auditar_base(base):
    padrao_email = r"^[^@\s]+@[^@\s]+\.[^@\s]+$"

    emails_problematicos = (
        base["email"].isna()
        | ~base["email"].astype("string").str.strip().str.match(
            padrao_email, na=False
        )
    ).sum()

    return {
        "Linhas": len(base),
        "Duplicatas": base.duplicated().sum(),
        "E-mails problemáticos": emails_problematicos,
        "Categorias únicas": base["categoria"].nunique(),
    }
```

```python
antes = auditar_base(df)
depois = auditar_base(df_clean)

comparacao = pd.DataFrame({
    "Métrica": antes.keys(),
    "Antes": antes.values(),
    "Depois": depois.values()
})
```
